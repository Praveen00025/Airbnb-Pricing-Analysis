```python
# Top 15 most reviewed Airbnb listings

most_reviewed = (
    df_clean[
        ['name', 'neighbourhood', 'room_type', 'price', 'number_of_reviews']
    ]
    .sort_values('number_of_reviews', ascending=False)
    .head(15)
)

most_reviewed
```

```python
plt.figure(figsize=(12, 7))

plt.barh(
    most_reviewed['name'].astype(str).str[:30],
    most_reviewed['number_of_reviews']
)

plt.xlabel('Number of Reviews')
plt.ylabel('Listing')
plt.title('Top 15 Most Reviewed Airbnb Listings')

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
# Create review groups

df_clean['review_group'] = pd.cut(
    df_clean['number_of_reviews'],
    bins=[-1, 10, 50, 100, 250, 500, float('inf')],
    labels=[
        '0-10',
        '11-50',
        '51-100',
        '101-250',
        '251-500',
        '500+'
    ]
)

review_price = (
    df_clean.groupby('review_group', observed=True)['price']
    .agg(['count', 'mean', 'median'])
)

review_price
```

```python
plt.figure(figsize=(10, 6))

review_price['mean'].plot(kind='bar')

plt.title('Average Airbnb Price by Number of Reviews')
plt.xlabel('Number of Reviews')
plt.ylabel('Average Price')
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
# Price vs Listing Count

comparison = (
    df_clean.groupby('neighbourhood')
    .agg(
        listing_count=('id', 'count'),
        average_price=('price', 'mean')
    )
    .sort_values('listing_count', ascending=False)
    .head(15)
)

comparison
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    comparison['listing_count'],
    comparison['average_price'],
    s=100
)

for neighbourhood, row in comparison.iterrows():
    plt.annotate(
        neighbourhood,
        (row['listing_count'], row['average_price']),
        xytext=(5, 5),
        textcoords='offset points'
    )

plt.xlabel('Number of Listings')
plt.ylabel('Average Price')
plt.title('Neighbourhood: Listing Count vs Average Price')

plt.tight_layout()
plt.show()
```

```python
# Top 10 neighbourhoods by total reviews

top_review_neighbourhoods = (
    df_clean.groupby('neighbourhood')['number_of_reviews']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_review_neighbourhoods
```

```python
plt.figure(figsize=(12, 6))

top_review_neighbourhoods.sort_values().plot(kind='barh')

plt.title('Top 10 Neighbourhoods by Total Reviews')
plt.xlabel('Total Number of Reviews')
plt.ylabel('Neighbourhood')

plt.tight_layout()
plt.show()
```

```python
# Neighbourhood performance

neighbourhood_performance = (
    df_clean.groupby('neighbourhood')
    .agg(
        listing_count=('id', 'count'),
        average_price=('price', 'mean'),
        average_reviews=('number_of_reviews', 'mean')
    )
)

# Minimum 50 listings wale neighbourhoods ko consider karenge
neighbourhood_performance = neighbourhood_performance[
    neighbourhood_performance['listing_count'] >= 50
]

neighbourhood_performance.sort_values(
    'average_reviews',
    ascending=False
).head(15)
```

```python
neighbourhood_performance['value_score'] = (
    neighbourhood_performance['average_reviews'] /
    neighbourhood_performance['average_price']
)

best_value = neighbourhood_performance.sort_values(
    'value_score',
    ascending=False
).head(15)

best_value
```

```python
plt.figure(figsize=(12, 7))

best_value['value_score'].sort_values().plot(kind='barh')

plt.title('Top 15 Best-Value Airbnb Neighbourhoods')
plt.xlabel('Value Score (Average Reviews / Average Price)')
plt.ylabel('Neighbourhood')

plt.tight_layout()
plt.show()
```
